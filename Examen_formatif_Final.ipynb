### A4) Mini‑analyse YAML (bonus +2 pts)  
Considérez cet extrait :

```yaml
name: Python Tests
on: [push, pull_request]

jobs:
  test:
    runs-on: ubuntu-latest
    steps:
      - name: Cloner le dépôt
        uses: actions/checkout@v3
      - name: Installer Python
        uses: actions/setup-python@v4
        with:
          python-version: '3.10'
      - name: Installer les dépendances
        run: pip install -r requirements.txt
      - name: Lancer les tests
        run: pytest tests/
```

**B1.** À quoi sert `with: python-version: '3.10'` ? *(1 pt)*  
C'est un paramètre passé à l'action setup-python pour spécifier que la version 3.10 de Python doit être installée et configurée dans l'environnement du runner.

**B2.** À quoi sert `with: uses: actions/checkout@v3` ? *(1 pt)* 
Cette ligne demande au runner d'utiliser l'action officielle de GitHub pour cloner le code source du dépôt actuel dans le répertoire de travail du runner, afin que les étapes suivantes puissent accéder aux fichiers du projet.

Utilisez `pytest` ou `unittest`.


````python
import pytest

# La fonction à tester (corrigée pour être fonctionnelle en Python 3)
def somme_quoi_inferieurs(n):
    somme = 0
    # On utilise // pour un entier et +1 pour inclure n/2 dans la boucle
    for i in range(1, (n // 2) + 1):
        if n % i == 0:
            somme += i
    return somme

#  TESTS UNITAIRES

def test_nombre_premier():
    # Pour 7, le seul diviseur < 7 est 1. Somme attendue = 1
    assert somme_quoi_inferieurs(7) == 1

def test_nombre_parfait():
    # Pour 6, les diviseurs sont 1, 2, 3. Somme : 1+2+3 = 6
    assert somme_quoi_inferieurs(6) == 6

def test_cas_limite_un():
    # Pour 1, il n'y a aucun diviseur inférieur à lui-même. Somme = 0
    assert somme_quoi_inferieurs(1) == 0

def test_type_invalide():
    # Vérifie qu'une erreur est levée si on passe une chaîne de caractères
    with pytest.raises(TypeError):
        somme_quoi_inferieurs("abc")
````

### Travail demandé :
Ajoutez les instructions de test unitaire au code en utilisant la classe `TestCase`.

- Montrez qu’un test échoue à cause de l’erreur logique dans `addition`
- Proposez la correction de la fonction


````python
import unittest

# 1. Fonction avec erreur logique (fournie)
def addition(a, b):
    return a + b - 1  # L'erreur est le "- 1"

# 2. Classe de test héritant de unittest.TestCase
class TestAddition(unittest.TestCase):

    def test_addition_positive(self):
        # Ce test échouera : addition(2, 3) retourne 4 au lieu de 5
        self.assertEqual(addition(2, 3), 5)

    def test_addition_zeros(self):
        # Ce test échouera : addition(0, 0) retourne -1 au lieu de 0
        self.assertEqual(addition(0, 0), 0)

    def test_addition_negatifs(self):
        # Ce test échouera : addition(-1, -1) retourne -3 au lieu de -2
        self.assertEqual(addition(-1, -1), -2)

# --- CORRECTION PROPOSÉE ---
def addition_corrigee(a, b):
    """Version corrigée de la fonction"""
    return a + b
````
